## Importación de librerias

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

from sklearn.datasets import load_wine

from sklearn.preprocessing import StandardScaler

from sklearn.decomposition import PCA
from sklearn.decomposition import FactorAnalysis
from sklearn.manifold import MDS
```

# Dataset Wine

Contiene características químicas de vinos

```python
wine = load_wine()

# DataFrame
df = pd.DataFrame(
    wine.data,
    columns=wine.feature_names
)

# Agregamos la clase del vino
df['target'] = wine.target
```

```python
df.head()
```

```python
print("\nDimensiones del dataset:")
print(df.shape)

print("\nColumnas:")
print(df.columns)

print("\nInformación general:")
print(df.info())

print("\nEstadísticas descriptivas:")
display(df.describe())
```

## Selección de variables

```python
X = df.drop('target', axis=1)
```

## Escalado de variables

Las técnicas de reducción de dimensionalidad son sensibles a las escalas.

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

## PCA

```python
pca = PCA(n_components=2)

# Transformación
X_pca = pca.fit_transform(X_scaled)
```

```python
X_pca
```

```python
print("\nShape original:")
print(X_scaled.shape)

print("\nShape luego de PCA:")
print(X_pca.shape)
```

### VARIANZA EXPLICADA

```python
print("\nVarianza explicada por componente:")

for i, varianza in enumerate(pca.explained_variance_ratio_):
    print(f"Componente {i+1}: {varianza:.4f}")

print("\nVarianza total explicada:")
print(pca.explained_variance_ratio_.sum())
```

### VISUALIZACIÓN PCA

```python
plt.scatter(
    X_pca[:,0],
    X_pca[:,1],
    c=wine.target
)

plt.xlabel("Componente Principal 1")
plt.ylabel("Componente Principal 2")

plt.title("PCA - Dataset Wine")

plt.colorbar(label="Clase")

plt.show()
```

- El dataset original tenía 13 variables.
- PCA las reduce a 2 componentes principales.
- Los componentes conservan la mayor varianza posible.
- Si aparecen grupos separados, PCA está capturando patrones importantes del dataset.

## ANÁLISIS FACTORIAL

```python
fa = FactorAnalysis(n_components=2)

X_fa = fa.fit_transform(X_scaled)
```

```python
X_fa.shape
```

### VISUALIZACIÓN FACTORIAL

```python
plt.scatter(
    X_fa[:,0],
    X_fa[:,1],
    c=wine.target
)

plt.xlabel("Factor 1")
plt.ylabel("Factor 2")

plt.title("Análisis Factorial - Dataset Wine")

plt.colorbar(label="Clase")

plt.show()
```

### CARGAS FACTORIALES

```python
cargas = pd.DataFrame(
    fa.components_,
    columns=X.columns,
    index=['Factor_1', 'Factor_2']
)

display(cargas)
```

- El análisis factorial intenta detectar factores ocultos.
- Las cargas factoriales indican cuánto influye cada variable en cada factor.
- Valores altos (positivos o negativos) indican fuerte relación con el factor.

## ESCALA MULTIDIMENSIONAL (MDS)

```python
mds = MDS(
    n_components=2,
    random_state=42
)

X_mds = mds.fit_transform(X_scaled)
```

```python
X_mds.shape
```

### VISUALIZACIÓN MDS

```python
plt.scatter(
    X_mds[:,0],
    X_mds[:,1],
    c=wine.target
)

plt.xlabel("Dimensión 1")
plt.ylabel("Dimensión 2")

plt.title("MDS - Dataset Wine")

plt.colorbar(label="Clase")

plt.show()
```

INTERPRETACIÓN:

- MDS intenta preservar distancias entre observaciones.
- Si dos puntos estaban cerca originalmente, deberían aparecer cercanos en el gráfico.
