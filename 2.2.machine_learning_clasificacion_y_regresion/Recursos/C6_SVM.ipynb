# Regresión por vectores de soporte

## Generación de datos sintéticos

```python
# Crearemos datos que sigan una función seno con algo de ruido aleatorio.
# Esto es ideal para mostrar la capacidad de SVR para manejar no linealidad.

# Generamos 100 puntos para X entre 0 y 2*pi
X = np.sort(5 * np.random.rand(100, 1), axis=0)
# Generamos y como una función seno de X, más un poco de ruido gausiano
y = np.sin(X).ravel() + np.random.randn(100) * 0.1
```

### Definición de los modelos SVR con Kernel: RBF, Lineal y Polinomial

Vamos a entrenar tres modelos SVR diferentes para comparar el efecto de los kernels.

```python
# Hiperparámetros comunes para una demostración justa.
# 'C' es el parámetro de regularización.
C = 1000

# Modelo 1: SVR con kernel RBF (Radial Basis Function) - El más flexible y común
svr_rbf = SVR(kernel='rbf', C=C)

# Modelo 2: SVR con kernel Lineal
svr_linear = SVR(kernel='linear', C=C)

# Modelo 3: SVR con kernel Polinomial de grado 3
svr_poly = SVR(kernel='poly', C=C)

# Entrenamos los modelos con nuestros datos (X, y)
svr_rbf.fit(X, y)
svr_linear.fit(X, y)
svr_poly.fit(X, y)
```

### Generación de la predicción, visualización y evaluación de los modelos

```python
# Hacemos predicciones sobre el mismo rango de X para poder dibujar las curvas.
y_rbf = svr_rbf.predict(X)
y_linear = svr_linear.predict(X)
y_poly = svr_poly.predict(X)
```

```python
# Creación del gráfico
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(12, 8))

# Dibujar los datos originales como un diagrama de dispersión
ax.scatter(X, y, color='darkorange', label='Datos Originales', zorder=2)

# Dibujar las líneas de predicción de cada modelo
ax.plot(X, y_rbf, color='navy', lw=2, label='Modelo RBF')
ax.plot(X, y_linear, color='cornflowerblue', lw=2, label='Modelo Lineal')
ax.plot(X, y_poly, color='c', lw=2, label='Modelo Polinomial')

# Añadir títulos y etiquetas para mayor claridad
ax.set_title('Comparación de Kernels en Regresión por Vectores de Soporte (SVR)', fontsize=16)
ax.set_xlabel('Datos de Entrada (X)', fontsize=12)
ax.set_ylabel('Datos de Salida (y)', fontsize=12)
ax.legend(loc='upper right', fontsize=11)

# Mostrar el gráfico
plt.show()
```

### Evaluación

```python
# Modelo RBF
r2_rbf = r2_score(y, y_rbf)
mse_rbf = mean_squared_error(y, y_rbf)
print(f"\nModelo RBF:")
print(f"  - Coeficiente R²: {r2_rbf:.4f}")
print(f"  - Error Cuadrático Medio (ECM): {mse_rbf:.4f}")

# Modelo Lineal
r2_linear = r2_score(y, y_linear)
mse_linear = mean_squared_error(y, y_linear)
print(f"\nModelo Lineal:")
print(f"  - Coeficiente R²: {r2_linear:.4f}")
print(f"  - Error Cuadrático Medio (ECM): {mse_linear:.4f}")

# Modelo Polinomial
r2_poly = r2_score(y, y_poly)
mse_poly = mean_squared_error(y, y_poly)
print(f"\nModelo Polinomial (Grado 3):")
print(f"  - Coeficiente R²: {r2_poly:.4f}")
print(f"  - Error Cuadrático Medio (ECM): {mse_poly:.4f}")
```
