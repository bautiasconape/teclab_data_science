## Importación de librerias

```python
#manipulación de datos
import pandas as pd
import numpy as np

# visualización
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("seaborn-v0_8-whitegrid")
pd.set_option('display.max_columns', None)
```

## Carga de datos

```python
url = "https://raw.githubusercontent.com/ybifoundation/Dataset/main/EmployeeAttrition.csv"
df = pd.read_csv(url)

df.head()
```

```python
# tamaño
df.shape
```

```python
df.info()
```

```python
# Resumen estadístico de las variables numéricas
df.describe()
```

## Limpieza y control de calidad

```python
print("Duplicados:", df.duplicated().sum())
```

```python
print("Nulos por columna:\n", df.isnull().sum())
```

```python
# Eliminar columnas irrelevantes (si aplica)
df = df.drop(columns=["EmployeeCount", "Over18", "StandardHours"], errors='ignore')
```

## Análisis exploratorio (EDA)

Análisis general de la variable objetivo

```python
# Distribución de la variable 'Attrition'
df["Attrition"].value_counts()
```

```python
# Proporción de empleados que se fueron
df["Attrition"].value_counts(normalize=True)*100
```

```python
sns.countplot(data=df, x="Attrition")
plt.title("Proporción de empleados que dejaron la empresa")
plt.show()
```

📊 Aproximadamente el 16 % de los empleados abandonaron la empresa.
Esto será útil más adelante para entender la desbalanceada de clases (problema común en IA).

### Análisis univariado y bivariado

Distribución de edad

```python
sns.histplot(df["Age"], bins=20, kde=True)
plt.title("Distribución de edad")
```

Distribución de ingresos

```python
sns.histplot(df["MonthlyIncome"], bins=25, kde=True)
plt.title("Distribución de ingresos mensuales")
plt.xlabel("Ingreso mensual")
plt.ylabel("Cantidad de empleados")
plt.show()
```

Satisfacción laboral

```python
sns.countplot(data=df, x="JobSatisfaction", hue="Attrition")
plt.title("Satisfacción laboral y abandono")
plt.xlabel("Nivel de satisfacción (1= bajo, 4= alto)")
plt.show()
```

Relación entre salario y abandono

```python
sns.boxplot(data=df, x="Attrition", y="MonthlyIncome")
plt.title("Ingresos mensuales vs abandono")
```

Promedio de satisfacción laboral por departamento

```python
sns.barplot(data=df, x="Department", y="JobSatisfaction", hue="Attrition")
plt.title("Satisfacción laboral por departamento y abandono")
plt.xticks(rotation=45)
```

### Detección de outliers

Ingresos y abandono

### Correlaciones

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), cmap="Blues", annot=False)
plt.title("Matriz de correlación entre variables numéricas")
plt.show()
```

#### Convertir la variable categórica a numérica

Transformamos "Attrition" en una variable binaria (Yes=1, No=0), lo cual permite incluirla en análisis numéricos.

```python
df['AttritionFlag'] = df['Attrition'].map({'Yes': 1, 'No': 0})
```

```python
corr = df.corr(numeric_only=True).sort_values(by='AttritionFlag', ascending=False)
corr['AttritionFlag']
```

Las variables con valores de correlación más altos (positivos o negativos) están más relacionadas con la probabilidad de abandono.

Por ejemplo, podrías ver que MonthlyIncome o YearsAtCompany correlacionan negativamente con AttritionFlag → a mayor ingreso o antigüedad, menor abandono.

### Relaciones múltiples entre variables relevantes

```python
cols = ["Age", "MonthlyIncome", "JobSatisfaction", "YearsAtCompany"]
sns.pairplot(df[cols], corner=True)
plt.suptitle("Análisis de relaciones entre variables seleccionadas", y=1.02)
plt.show()
```

🧠 Este gráfico permite observar tendencias generales y posibles asociaciones entre variables.
